```python
# [CELL 4] PARAMETRIC PROCESSING
def user_summary(title, *scores, **info):
    print(f"=== {title} ===")
    total = sum(scores)
    average = total / len(scores)

    print("Scores:", scores)
    print("Total:", total)
    print("Average:", round(average, 2))

    for k, v in info.items():
        print(f"{k}: {v}")

    return average


avg = user_summary(
    f"{LAST_NAME} Academic Report",
    SEED_DIGIT * 10,
    )
```
